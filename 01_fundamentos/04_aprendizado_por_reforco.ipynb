# Aprendizado por reforço

O aprendizado por reforço é um tipo diferente de aprendizado de máquina no qual precisamos fazer com que o computador execute uma tarefa. Além de dados, o modelo recebe um ambiente e um agente que deve navegar nesse ambiente. O agente tem um objetivo ou um conjunto de objetivos.

O ambiente tem recompensas e punições que guiam o agente a tomar as decisões corretas para alcançar seu objetivo. Tudo isso soa um pouco abstrato, mas vamos ver um exemplo.



Na figura a seguir, vemos um mundo em grade com um robô no canto inferior esquerdo. Esse é o nosso agente. O objetivo é chegar ao baú do tesouro no canto superior direito da grade. Na grade, também podemos ver uma montanha, o que significa que não podemos passar por esse quadrado, porque o robô não pode escalar montanhas. Também vemos um dragão, que atacará o robô caso ele pouse em seu quadrado, o que significa que parte do nosso objetivo é não pousar lá. Este é o jogo. E para dar ao robô informações sobre como proceder, mantemos um registro de uma pontuação. A pontuação começa em zero. Se o robô chegar ao baú do tesouro, ganhamos 100 pontos. Se o robô alcançar o dragão, perdemos 50 pontos. E para garantir que nosso robô se mova rapidamente, podemos dizer que para cada passo que o robô der, perdemos 1 ponto, porque o robô perde energia enquanto anda.

A maneira de treinar esse algoritmo, em termos muito gerais, é a seguinte: O robô começa a andar, registrando sua pontuação e lembrando quais passos o levaram até lá. Em algum momento, ele pode encontrar o dragão, perdendo muitos pontos. Portanto, ele aprende a associar o quadrado do dragão e os quadrados próximos a ele com pontuações baixas. Em algum momento, ele também pode chegar ao baú do tesouro e aprende a começar a associar esse quadrado e os quadrados próximos a ele com pontuações altas. Depois de jogar esse jogo por um longo tempo, o robô terá uma boa ideia de quão bons são cada um dos quadrados e poderá seguir o caminho pelos quadrados até o baú do tesouro. ### Aplicações avançadas do aprendizado por reforço

O aprendizado por reforço tem inúmeras aplicações de ponta, incluindo as seguintes:

- **Jogos**: avanços recentes no ensino de computadores a vencer em jogos, como Go ou xadrez, usam aprendizado por reforço. Além disso, agentes foram ensinados a vencer em jogos de Atari, como Breakout ou Super Mario.
- **Robótica**: o aprendizado por reforço é amplamente utilizado para ajudar robôs a realizar tarefas como pegar caixas, limpar uma sala ou até mesmo dançar!
- **Carros autônomos**: técnicas de aprendizado por reforço são usadas para ajudar o carro a realizar muitas tarefas, como planejamento de rotas ou comportar-se em ambientes específicos.
