# Etapas no desenvolvimento de uma aplicação de aprendizado de máquina

**Objetivo deste notebook:** Relacionar as etapas de um projeto de Machine Learning, da definição do problema à avaliação do modelo.


Nossa abordagem para entender e desenvolver uma aplicação usando aprendizado de máquina seguirá um procedimento semelhante a este:

1. **Coletar dados**: Você pode coletar amostras extraindo dados de um site, obtendo informações de um feed RSS ou de uma API. Você pode ter um dispositivo que colete medições de velocidade do vento e as envie para você, ou níveis de glicose no sangue, ou qualquer coisa que possa ser medida. As opções são infinitas. Para economizar tempo e esforço, você pode usar dados disponíveis publicamente.

2. **Preparar os dados de entrada**: Uma vez que você tenha esses dados, precisa garantir que eles estejam em um formato utilizável. Você pode precisar fazer algum formatação específica do algoritmo aqui. Alguns algoritmos precisam de características em um formato especial, alguns algoritmos podem lidar com variáveis alvo e características como strings, e alguns precisam que sejam inteiros. Falaremos sobre isso mais tarde, mas a formatação específica do algoritmo geralmente é trivial em comparação com a coleta de dados.

3. **Analisar os dados de entrada**: Isso envolve olhar os dados da tarefa anterior. Isso pode ser tão simples quanto olhar os dados que você analisou para garantir que os passos 1 e 2 estão realmente funcionando e que você não tem um monte de valores vazios. Você também pode olhar os dados para ver se reconhece algum padrão ou se há algo óbvio, como alguns pontos de dados que são muito diferentes do resto do conjunto. Plotar dados em uma, duas ou três dimensões também pode ajudar. Mas, na maioria das vezes, você terá mais de três características, e não pode facilmente plotar os dados em todas as features ao mesmo tempo. No entanto, você pode usar alguns métodos para reduzir múltiplas dimensões em duas ou três para visualizar os dados.

4. **Verificar os dados**: Se você está trabalhando com um sistema de produção e sabe como os dados devem se parecer, ou confia na fonte deles, pode pular esta etapa. Esta etapa requer envolvimento humano, e para um sistema automatizado, você não quer envolvimento humano. O valor desta etapa é garantir que você não tenha lixo entrando no sistema.

5. **Treinar o algoritmo**: É aqui que o aprendizado de máquina ocorre. Este passo e o próximo são onde residem os “núcleos” dos algoritmos, dependendo do algoritmo. Você alimenta o algoritmo com dados bons e limpos dos primeiros dois passos e extrai conhecimento ou informação. Este conhecimento geralmente é armazenado em um formato que é prontamente utilizável por uma máquina para os próximos dois passos. No caso do aprendizado não supervisionado, não há etapa de treinamento porque você não tem um valor alvo. Tudo é usado na próxima etapa.

6. **Testar o algoritmo**: É aqui que a informação aprendida na etapa anterior é colocada em uso. Quando você está avaliando um algoritmo, testa para ver quão bem ele se sai. No caso do aprendizado supervisionado, você tem alguns valores conhecidos que pode usar para avaliar o algoritmo. No aprendizado não supervisionado, pode ser necessário usar algumas outras métricas para avaliar o sucesso. Em qualquer caso, se você não estiver satisfeito, pode voltar ao passo 4, mudar algumas coisas e tentar testar novamente. Frequentemente, a coleta ou preparação dos dados pode ter sido o problema, e você terá que voltar ao passo 1.

7. **Usar o algoritmo**: Aqui você cria um programa real para executar alguma tarefa e, mais uma vez, verifica se todos os passos anteriores funcionaram como esperado. Você pode encontrar alguns dados novos e precisar revisitar os passos 1-5.
