# Aprendizado supervisionado

**Objetivo deste notebook:** Identificar problemas de aprendizado supervisionado e distinguir tarefas de regressão e classificação.


Neste tipo de aprendizado, os algoritmos são treinados em um conjunto de dados rotulados, onde cada exemplo de treinamento é uma entrada com um rótulo associado. O objetivo é aprender uma função que mapeia as entradas para os rótulos, de modo que o modelo possa fazer previsões precisas sobre novos dados. Exemplos de aprendizado supervisionado incluem regressão linear, regressão logística, árvores de decisão e redes neurais.

Lembrando de outros momentos do nosso curso:

- **Dados numéricos** são qualquer tipo de dado que usa números, como 4, 2.35 ou -199. Exemplos de dados numéricos são preços, tamanhos ou pesos.
- **Dados categóricos** são qualquer tipo de dado que usa categorias ou estados, como masculino/feminino ou gato/cachorro/pássaro. Para esse tipo de dado, temos um conjunto finito de categorias para associar a cada ponto de dado.

Isso dá origem aos seguintes dois tipos de modelos de aprendizado supervisionado:

- **Modelos de regressão** são os tipos de modelos que predizem dados numéricos. A saída de um modelo de regressão é um número, como o peso do animal.
- **Modelos de classificação** são os tipos de modelos que predizem dados categóricos. A saída de um modelo de classificação é uma categoria ou um estado, como o tipo de animal (gato ou cachorro).

Vamos olhar para dois exemplos de modelos de aprendizado supervisionado, um de regressão e um de classificação.

- **Modelo 1: modelo de preços de casas (regressão)**. Nesse modelo, cada ponto de dado é uma casa. O rótulo de cada casa é o seu preço. Nosso objetivo é que, quando uma nova casa (ponto de dado) entrar no mercado, possamos prever seu rótulo, ou seja, seu preço.
- **Modelo 2: modelo de detecção de spam em e-mails (classificação)**. Nesse modelo, cada ponto de dado é um e-mail. O rótulo de cada e-mail é spam ou não spam. Nosso objetivo é que, quando um novo e-mail (ponto de dado) chegar à nossa caixa de entrada, possamos prever seu rótulo, ou seja, se é spam ou não.

Note a diferença entre os modelos 1 e 2:
- O modelo de preços de casas é um modelo que pode retornar um número entre muitas possibilidades, como $100, $250.000 ou $3.125.672,33. Assim, é um modelo de regressão.
- O modelo de detecção de spam, por outro lado, pode retornar apenas duas coisas: spam ou não spam. Assim, é um modelo de classificação.


## Modelos de regressão

Como discutimos anteriormente, modelos de regressão são aqueles em que o rótulo que queremos prever é um número. Esse número é previsto com base nas features. No exemplo de casas, as features podem ser qualquer coisa que descreva uma casa, como o tamanho, o número de cômodos, a distância até a escola mais próxima ou a taxa de criminalidade no bairro.

Outros lugares onde se pode usar modelos de regressão incluem:

- **Mercado de ações**: prevendo o preço de uma determinada ação com base em outros preços de ações e outros sinais de mercado.
- **Medicina**: prevendo a expectativa de vida de um paciente ou o tempo esperado de recuperação, com base nos sintomas e no histórico médico do paciente.
- **Vendas**: prevendo a quantidade esperada de dinheiro que um cliente gastará, com base nos dados demográficos e no comportamento de compra anterior do cliente.
- **Recomendações de vídeos**: prevendo a quantidade esperada de tempo que um usuário assistirá a um vídeo, com base nos dados demográficos do usuário e em outros vídeos que ele assistiu.

O método mais comum usado para regressão é a regressão linear, que utiliza funções lineares (linhas ou objetos semelhantes) para fazer nossas previsões com base nas features. Outros métodos populares usados para regressão são a regressão de árvore de decisão, e métodos de conjunto, como florestas aleatórias (random forests), AdaBoost, árvores de decisão aumentadas por gradiente (gradient boosted trees) e XGBoost.

## Modelos de classificação

**Modelos de classificação** são aqueles em que o rótulo que queremos prever é um estado pertencente a um conjunto finito de estados. Os modelos de classificação mais comuns predizem um "sim" ou "não", mas muitos outros modelos usam um conjunto maior de estados.

No exemplo de reconhecimento de spam em e-mails, o modelo prevê o estado do e-mail (ou seja, spam ou não spam) a partir das características do e-mail. Nesse caso, as características do e-mail podem ser as palavras nele contidas, o número de erros ortográficos, o remetente ou qualquer outra coisa que descreva o e-mail.

Outra aplicação comum de classificação é o reconhecimento de imagens. Os modelos de reconhecimento de imagens mais populares têm como entrada os pixels da imagem e geram uma previsão do que a imagem representa. Dois dos conjuntos de dados mais famosos para reconhecimento de imagens são MNIST e CIFAR-10. MNIST contém aproximadamente 60.000 imagens em preto e branco de 28 por 28 pixels de dígitos manuscritos, que são rotulados de 0 a 9. Essas imagens vêm de uma combinação de fontes, incluindo o American Census Bureau e um repositório de dígitos manuscritos por estudantes do ensino médio americanos. O conjunto de dados CIFAR-10 contém 60.000 imagens coloridas de 32 por 32 pixels de coisas diferentes. Essas imagens são rotuladas com 10 objetos diferentes (daí o 10 no nome), a saber, aviões, carros, pássaros, gatos, cervos, cachorros, sapos, cavalos, navios e caminhões. Esta base de dados é mantida pelo Instituto Canadense de Pesquisas Avançadas (CIFAR). Ambas as bases serão estudadas em módulos futuros.

Algumas outras aplicações poderosas de modelos de classificação são:

- **Análise de sentimentos**: prever se uma crítica de filme é positiva ou negativa, com base nas palavras da crítica.
- **Tráfego de sites**: prever se um usuário clicará em um link ou não, com base nos dados demográficos do usuário e na interação anterior com o site.
- **Mídias sociais**: prever se um usuário fará amizade ou interagirá com outro usuário, com base nos dados demográficos, histórico e amigos em comum.
- **Recomendações de vídeos**: prever se um usuário assistirá a um vídeo, com base nos dados demográficos do usuário e em outros vídeos que ele assistiu.

Existem diferentes métodos para classificação, incluindo perceptrons, classificadores logísticos, árvores de decisão, redes neurais, máquinas de vetor de suporte e métodos de conjunto.
