# Resumo

**Objetivo deste notebook:** Revisar os conceitos da introdução antes de avançar para os experimentos de regressão.


- Existem vários tipos de aprendizado de máquina, incluindo aprendizado supervisionado, aprendizado não supervisionado e aprendizado por reforço.
- Os dados podem ser rotulados ou não rotulados. Dados rotulados contêm uma feature especial, ou rótulo, que buscamos prever. Dados não rotulados não contêm essa feature.
- O aprendizado supervisionado é usado em dados rotulados e consiste na construção de modelos que prevêem os rótulos para dados não vistos.
- O aprendizado não supervisionado é usado em dados não rotulados e consiste em algoritmos que simplificam nossos dados sem perder muita informação. O aprendizado não supervisionado é frequentemente usado como uma etapa de pré-processamento.
- Dois tipos comuns de algoritmos de aprendizado supervisionado são chamados de regressão e classificação.
  - Modelos de regressão são aqueles em que a resposta é um número.
  - Modelos de classificação são aqueles em que a resposta é de um tipo ou classe.
- Dois tipos comuns de algoritmos de aprendizado não supervisionado são o agrupamento (clustering) e a redução de dimensionalidade.
  - O agrupamento é usado para agrupar dados em clusters semelhantes para extrair informações ou facilitar o manuseio.
  - A redução de dimensionalidade é uma maneira de simplificar nossos dados, unindo certas características semelhantes e perdendo o mínimo de informação possível.
- O aprendizado de máquina generativo é um tipo de aprendizado não supervisionado, que consiste em gerar dados que são semelhantes ao nosso conjunto de dados. Modelos generativos podem gerar rostos realistas, compor músicas e escrever poesias.
- O aprendizado por reforço é um tipo de aprendizado de máquina em que um agente deve navegar em um ambiente e alcançar um objetivo. É amplamente utilizado em muitas aplicações de ponta.
- Usaremos o Scikit-learn, uma biblioteca de aprendizado de máquina em Python, para implementar algoritmos de aprendizado de máquina.
