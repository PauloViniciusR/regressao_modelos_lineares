# O que é Machine Learning?

Machine Learning é um subcampo da Inteligência Artificial (IA) que se concentra no desenvolvimento de algoritmos e técnicas que permitem que as máquinas aprendam a partir de dados. Em outras palavras, o Machine Learning é o processo de ensinar uma máquina a realizar uma tarefa sem ser explicitamente programada para fazê-lo.



Para facilitar, vamos dar algumas definições:

**Inteligência Artificial (IA)**:

**O que é**: É o campo da ciência que busca criar sistemas capazes de realizar tarefas que normalmente requerem inteligência humana. Isso inclui entender a linguagem, reconhecer padrões, resolver problemas e tomar decisões.

**Exemplo 1**: Um jogo de xadrez que pode jogar contra um humano. O programa analisa as jogadas possíveis e escolhe a melhor estratégia para vencer.

**Exemplo 2**: Sistemas de atendimento ao cliente que utilizam chatbots para responder perguntas comuns. O chatbot entende perguntas simples e fornece respostas automáticas.

**Analogia**: Pense na IA como um grande guarda-chuva que cobre qualquer tipo de "inteligência" que uma máquina possa ter. Isso inclui entender comandos de voz, reconhecer imagens, jogar jogos, etc.

---

**Machine Learning (ML)**:

**O que é**: É uma abordagem dentro da IA que permite às máquinas aprenderem com dados em vez de serem explicitamente programadas. Elas identificam padrões e fazem previsões com base nos dados fornecidos.

**Exemplo 1**: Filtragem de spam no e-mail. O sistema aprende a partir de exemplos de e-mails marcados como spam e não spam e, com o tempo, se torna melhor em identificar automaticamente quais e-mails são indesejados.

**Exemplo 2**: Previsão de vendas. Um sistema de ML pode analisar dados históricos de vendas para prever a demanda futura de produtos.

**Analogia**: Imagine ensinar um amigo a distinguir gatos de cachorros mostrando muitas fotos de cada um. Com o tempo, seu amigo aprenderá a reconhecer as diferenças e fazer essa distinção sozinho.

---

**Deep Learning (DL)**:

**O que é**: É uma técnica de ML que utiliza redes neurais artificiais com muitas camadas (daí o termo "profundo") para processar grandes quantidades de dados complexos. É especialmente útil para tarefas como reconhecimento de imagens e processamento de linguagem natural.

**Exemplo 1**: Reconhecimento de voz em assistentes virtuais como Alexa e Google Home. O sistema analisa a voz do usuário e transforma o som em texto, reconhecendo comandos específicos.

**Exemplo 2**: Diagnóstico médico automatizado. Redes neurais profundas podem analisar imagens médicas, como radiografias, para detectar sinais de doenças com alta precisão.

**Analogia**: Pense em DL como ensinar um aluno avançado que, em vez de apenas distinguir gatos de cachorros, consegue reconhecer raças específicas, identificar detalhes como a posição dos animais na foto e até distinguir entre diferentes expressões faciais dos animais. Tudo isso é possível graças ao treinamento intensivo com muitas camadas de análise.

---

**Resumindo**:

1. **Inteligência Artificial**: É o campo geral que busca criar máquinas inteligentes.
   - **Exemplo**: Um robô que pode andar e evitar obstáculos.

2. **Machine Learning**: É uma técnica de IA onde as máquinas aprendem com dados.
   - **Exemplo**: Um aplicativo que recomenda músicas com base nas músicas que você já ouviu.

3. **Deep Learning**: É um método de ML que utiliza redes neurais com muitas camadas para analisar dados complexos.
   - **Exemplo**: Um sistema que pode identificar tumores em imagens de ressonância magnética.

**Resumo Visual**:
- **IA**: O guarda-chuva que cobre todas as formas de inteligência de máquina.
  - **ML**: Uma parte da IA que aprende com dados.
    - **DL**: Uma técnica de ML que usa redes neurais profundas para aprender de forma avançada.

## Aplicações de Machine Learning

Machine learning tem aplicações em muitos campos, como os seguintes:

- Prever preços de casas com base no tamanho da casa, número de cômodos e localização.
- Prever os preços do mercado de ações de hoje com base nos preços de ontem e outros fatores do mercado.
- Detectar e-mails de spam e não spam com base nas palavras no e-mail e no remetente.
- Reconhecer imagens como rostos ou animais, com base nos pixels da imagem.
- Diagnosticar pacientes como doentes ou saudáveis.
- Segmentar o mercado em grupos semelhantes com base na localização, poder aquisitivo e interesses.
- Processar documentos longos de texto e gerar um resumo.
- Recomendar vídeos ou filmes a um usuário (por exemplo, no YouTube ou Netflix).
- Treinar carros autônomos para navegar por uma cidade sozinhos.

## Tomada de decisão por Machine Learning

Machine Learning é o processo de ensinar uma máquina a realizar uma tarefa sem ser explicitamente programada para fazê-lo.

Com os dados, as máquinas constroem um modelo matemático que captura os padrões nos dados. Esse modelo é então usado para fazer previsões ou tomar decisões sobre novos dados.

Um **modelo** é uma representação matemática dos padrões nos dados que pode ser usada para tomada de decisão. Por exemplo, um modelo de previsão de preços de casas pode ser uma equação que leva em consideração o tamanho da casa, o número de cômodos e a localização. Esse modelo pode ser usado para prever o preço de uma nova casa com base nessas características.

Um **algoritmo** é um conjunto de regras e procedimentos que a máquina segue para aprender com os dados e construir um modelo. Existem muitos tipos diferentes de algoritmos de Machine Learning, cada um com suas próprias vantagens e desvantagens.

### Alguns termos importantes que vamos precisar

#### Features (Características, Atributos, Variáveis)

Desde nosso módulo de Pandas, definimos variáveis como as propriedades ou atributos dos dados. Se nossos dados estão em uma tabela, as características são as colunas da tabela. Considerando um exemplo de animais de estimação, as características podem ser tamanho, nome, tipo ou peso. As características podem até ser as cores dos pixels em uma imagem do animal. Isso é o que descreve nossos dados. Algumas características são especiais, e nós as chamamos de rótulos.

#### Rótulos

Este conceito é um pouco menos direto, porque depende do contexto do problema que estamos tentando resolver. Normalmente, se estamos tentando prever uma característica específica com base nas outras, essa característica é o rótulo, ou variável alvo. Se estamos tentando prever o tipo de animal (por exemplo, gato ou cachorro) com base nas informações desse animal, então o rótulo é o tipo de animal (gato/cachorro). Se estamos tentando prever se o animal está doente ou saudável com base em sintomas e outras informações, então o rótulo é o estado do animal (doente/saudável). Se estamos tentando prever a idade do animal, então o rótulo é a idade (um número).

#### Previsões

Temos usado livremente o conceito de fazer previsões, mas agora vamos esclarecer. O objetivo de um modelo preditivo de aprendizado de máquina é adivinhar os rótulos nos dados. A adivinhação que o modelo faz é chamada de previsão.

#### Dados rotulados e não rotulados

Agora que sabemos o que são rótulos, podemos entender que existem dois tipos principais de dados: dados rotulados e não rotulados.

- **Dados Rotulados**: Dados que vêm com rótulos. Um exemplo de dados rotulados é um conjunto de dados de e-mails que vem com uma coluna que registra se os e-mails são spam ou não, ou uma coluna que registra se o e-mail é relacionado ao trabalho.

- **Dados Não Rotulados**: Dados que vêm sem rótulos. Um exemplo de dados não rotulados é um conjunto de dados de e-mails que não tem uma coluna específica que estamos interessados em prever.

## Tipos de Machine Learning

Os algoritmos de Machine Learning podem ser divididos em três categorias principais, com base no tipo de aprendizado que realizam:

1. **Aprendizado Supervisionado**: Neste tipo de aprendizado, os algoritmos são treinados em um conjunto de dados rotulados, onde cada exemplo de treinamento é uma entrada com um rótulo associado. O objetivo é aprender uma função que mapeia as entradas para os rótulos, de modo que o modelo possa fazer previsões precisas sobre novos dados. Exemplos de aprendizado supervisionado incluem regressão linear, regressão logística, árvores de decisão e redes neurais.
2. **Aprendizado Não Supervisionado**: Neste tipo de aprendizado, os algoritmos são treinados em um conjunto de dados não rotulados, onde não há rótulos associados a cada exemplo de treinamento. O objetivo é encontrar padrões ou estrutura nos dados, como agrupamentos ou associações, sem a necessidade de rótulos. Exemplos de aprendizado não supervisionado incluem agrupamento k-means, análise de componentes principais (PCA) e regras de associação.
3. **Aprendizado por Reforço**: Neste tipo de aprendizado, os algoritmos aprendem a partir da interação com um ambiente, recebendo feedback positivo ou negativo com base em suas ações. O objetivo é aprender uma política que maximize uma recompensa cumulativa ao longo do tempo. Exemplos de aprendizado por reforço incluem Q-learning, aprendizado profundo por reforço e algoritmos genéticos.
